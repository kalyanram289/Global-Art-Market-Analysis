```python
import pandas as pd

file_path = '/content/df_for_ml_improved_up_to_2012.csv'


df = pd.read_csv(file_path)

print("DataFrame loaded successfully.")
print("Columns in the DataFrame:")
print(df.columns.tolist())
```

```python
keep_cols = [
    'case_id',
    'year',
    'height',
    'width',
    'size_inchsqr',
    'price_usd',

    'ranking',
    'age',

    'estimate_min_usd',
    'estimate_max_usd',
    'estimate_center_usd',
    'estimate_range_usd',

    'Country - Austria',
    'Country - China',
    'Country - Dutch',
    'Country - France',
    'Country - Germany',
    'Country - Japan',
    'Country - Others',
    'Country - Swiss',
    'Country - UK',
    'Country - USA',

    'Continent - East Asia',
    'Continent - Europe',
    'Continent - North America',
    'Continent - Others',

    'Tier - 1',
    'Tier - 2',
    'Tier - 3',
    'Tier - 4',

    'Genre___Photography',
    'Genre___Print',
    'Genre___Sculpture',
    'Genre___Painting',
    'Genre___Others',

    'eventdate'
]

art = df[keep_cols].copy()

print(art.shape)
art.head()
```

```python
country_cols = [
    'Country - Austria',
    'Country - China',
    'Country - Dutch',
    'Country - France',
    'Country - Germany',
    'Country - Japan',
    'Country - Others',
    'Country - Swiss',
    'Country - UK',
    'Country - USA'
]

def get_country(row):
    for col in country_cols:
        if row[col] == 1:
            return col.replace('Country - ', '')
    return 'Unknown'

art['Country'] = art.apply(get_country, axis=1)
```

```python
art['Country'].value_counts()
```

```python
continent_cols = [
    'Continent - East Asia',
    'Continent - Europe',
    'Continent - North America',
    'Continent - Others'
]

def get_continent(row):
    for col in continent_cols:
        if row[col] == 1:
            return col.replace('Continent - ', '')
    return 'Unknown'

art['Continent'] = art.apply(get_continent, axis=1)

art['Continent'].value_counts()
```

```python
genre_cols = [
    'Genre___Photography',
    'Genre___Print',
    'Genre___Sculpture',
    'Genre___Painting',
    'Genre___Others'
]

def get_genre(row):
    for col in genre_cols:
        if row[col] == 1:
            return col.replace('Genre___', '')
    return 'Unknown'

art['Genre'] = art.apply(get_genre, axis=1)

art['Genre'].value_counts()
```

```python
tier_cols = [
    'Tier - 1',
    'Tier - 2',
    'Tier - 3',
    'Tier - 4'
]

def get_tier(row):
    for col in tier_cols:
        if row[col] == 1:
            return col.replace('Tier - ', 'Tier ')
    return 'Unknown'

art['Tier'] = art.apply(get_tier, axis=1)

art['Tier'].value_counts()
```

```python
drop_cols = country_cols + continent_cols + genre_cols + tier_cols

art = art.drop(columns=drop_cols)

print(art.columns.tolist())
```

```python
import numpy as np

art['estimate_status'] = np.select(
    [
        art['price_usd'] < art['estimate_min_usd'],
        art['price_usd'] > art['estimate_max_usd']
    ],
    [
        'Below Estimate',
        'Above Estimate'
    ],
    default='Within Estimate'
)
```

```python
art['estimate_status'].value_counts()
```

```python
art.groupby('estimate_status')['price_usd'].agg(
    ['count', 'mean', 'median']
)
```

```python
art['estimate_ratio'] = (
    art['price_usd'] / art['estimate_center_usd']
)

art['estimate_ratio'].describe()
```

```python
country_estimate = art.groupby('Country').agg(
    artworks=('case_id', 'count'),
    median_price=('price_usd', 'median'),
    median_estimate_ratio=('estimate_ratio', 'median'),
    above_estimate_pct=('estimate_status',
                       lambda x: (x == 'Above Estimate').mean() * 100)
).sort_values('median_estimate_ratio', ascending=False)

country_estimate
```

```python
tier_estimate = art.groupby('Tier').agg(
    artworks=('case_id', 'count'),
    median_price=('price_usd', 'median'),
    median_estimate_ratio=('estimate_ratio', 'median'),
    above_estimate_pct=('estimate_status',
                       lambda x: (x == 'Above Estimate').mean() * 100)
).sort_values('median_estimate_ratio', ascending=False)

tier_estimate
```
